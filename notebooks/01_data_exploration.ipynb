## 3. Load CICIDS2017 Dataset

```python
# CICIDS2017 files
cicids_path = 'data/raw/cicids2017/'

cicids_files = [
    'Monday-WorkingHours.pcap_ISCX.csv',
    'Tuesday-WorkingHours.pcap_ISCX.csv',
    'Wednesday-workingHours.pcap_ISCX.csv',
    'Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv',
    'Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX.csv',
    'Friday-WorkingHours-Morning.pcap_ISCX.csv',
    'Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX.csv',
    'Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv'
]

print('📂 CICIDS2017 expected files:')
for f in cicids_files:
    print(f'   - {f}')
```

```python
def load_cicids2017(path):
    """
    Load and combine all CICIDS2017 CSV files
    """
    dfs = []
    
    for file in os.listdir(path):
        if file.endswith('.csv'):
            print(f'📖 Loading: {file}')
            df = pd.read_csv(os.path.join(path, file), low_memory=False)
            dfs.append(df)
            print(f'   Shape: {df.shape}')
    
    if dfs:
        combined = pd.concat(dfs, ignore_index=True)
        print(f'\n✅ Combined shape: {combined.shape}')
        return combined
    else:
        print('❌ No CSV files found!')
        return None

# Load dataset (uncomment when files are downloaded)
# cicids_df = load_cicids2017(cicids_path)
```

## 5. Basic Data Exploration

```python
def explore_dataset(df, name):
    """
    Basic exploration of dataset
    """
    print(f'\n{"="*50}')
    print(f'📊 Dataset: {name}')
    print(f'{"="*50}')
    
    print(f'\n📏 Shape: {df.shape}')
    print(f'   - Rows: {df.shape[0]:,}')
    print(f'   - Columns: {df.shape[1]}')
    
    print(f'\n📋 Data Types:')
    print(df.dtypes.value_counts())
    
    print(f'\n❓ Missing Values:')
    missing = df.isnull().sum()
    missing_pct = (missing / len(df)) * 100
    missing_df = pd.DataFrame({'Missing': missing, 'Percentage': missing_pct})
    print(missing_df[missing_df['Missing'] > 0])
    
    print(f'\n🏷️ Label Distribution:')
    if 'Label' in df.columns:
        print(df['Label'].value_counts())
    elif 'label' in df.columns:
        print(df['label'].value_counts())
    
    return df.describe()

# Explore datasets (uncomment when loaded)
# explore_dataset(cicids_df, 'CICIDS2017')
# explore_dataset(ciciot_df, 'CIC-IoT2023')
```

## 6. Visualize Attack Distribution

```python
def plot_label_distribution(df, name, label_col='Label'):
    """
    Plot attack type distribution
    """
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Bar plot
    label_counts = df[label_col].value_counts()
    ax1 = axes[0]
    label_counts.plot(kind='bar', ax=ax1, color='steelblue', edgecolor='black')
    ax1.set_title(f'{name} - Attack Distribution', fontsize=12, fontweight='bold')
    ax1.set_xlabel('Attack Type')
    ax1.set_ylabel('Count')
    ax1.tick_params(axis='x', rotation=45)
    
    # Pie chart
    ax2 = axes[1]
    label_counts.plot(kind='pie', ax=ax2, autopct='%1.1f%%')
    ax2.set_title(f'{name} - Attack Percentage', fontsize=12, fontweight='bold')
    ax2.set_ylabel('')
    
    plt.tight_layout()
    plt.savefig(f'results/figures/{name}_distribution.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    print(f'\n✅ Figure saved: results/figures/{name}_distribution.png')

# Plot distributions (uncomment when loaded)
# plot_label_distribution(cicids_df, 'CICIDS2017')
# plot_label_distribution(ciciot_df, 'CIC-IoT2023', 'label')
```
